## Outlier handling using IQR method

```python
plt.boxplot(life_data['Population'])
plt.show()
```

```python
plt.boxplot(life_data['Adult Mortality'])
plt.show()
```

```python
plt.boxplot(life_data['infant deaths'])
plt.show()
```

```python
for col in num_cols:
    Q1 = life_data[col].quantile(0.25)
    Q3 = life_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((life_data[col] < lower) | (life_data[col] > upper)).sum()

    print(col, ":", outliers)
```

### cliping method

```python
# Target column ozhivakki numerical columns edukkuka
num_cols = life_data.drop('Life expectancy ', axis=1).select_dtypes(include='number').columns

# IQR method + Capping
for col in num_cols:
    Q1 = life_data[col].quantile(0.25)
    Q3 = life_data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    life_data[col] = life_data[col].clip(lower=lower, upper=upper)
```

```python
plt.figure(figsize=(15, 8))
sns.boxplot(data=life_data[num_cols])
plt.xticks(rotation=90)
plt.show()
```

# Seperate Feature and Target Colunm

```python
x = life_data.drop("Life expectancy ", axis=1) ### Traget Colunms
y = life_data["Life expectancy "]
```

```python
# Target variable
y = life_data["Life expectancy "].reset_index(drop=True)

# Feature variables
x = life_data.drop("Life expectancy ", axis=1).reset_index(drop=True)
```

#Scaling

```python
# creating a copy of my features
x1=x.copy()
```

#Train Test Split

```python
# scaled x_train1 and x_test1
from sklearn.model_selection import train_test_split
x_train1, x_test1, y_train, y_test = train_test_split(x1, y, test_size=0.2, random_state=42,)

```

```python
# non scaled x_train and x_test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
for i in x1.columns:
  x1[i] = scaler.fit_transform(x1[[i]])
```

#Model

##Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Linear regression with scaled data
lr_model_scaled = LinearRegression()
lr_model_scaled.fit(x_train1, y_train)
y_pred_lr_model_scaled = lr_model_scaled.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_lr_model_scaled))
print('MAE is: ', mean_absolute_error(y_test, y_pred_lr_model_scaled))
print('RMSE is: ', mean_squared_error(y_test, y_pred_lr_model_scaled) ** 0.5)
```

#Decision Tree

```python


# Decision Tree Regression with non scalled data

from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Decision Tree Regression
dt_model = DecisionTreeRegressor(random_state=42)

dt_model.fit(x_train1, y_train)
y_pred_dt = dt_model.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_dt))
print('MAE is: ', mean_absolute_error(y_test, y_pred_dt))
print('RMSE is: ', mean_squared_error(y_test, y_pred_dt) ** 0.5)
```

#Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Random Forest Regression
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

rf_model.fit(x_train1, y_train)
y_pred_rf = rf_model.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_rf))
print('MAE is: ', mean_absolute_error(y_test, y_pred_rf))
print('RMSE is: ', mean_squared_error(y_test, y_pred_rf) ** 0.5)
```

#Lasso

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Lasso Regression with scaled data
lasso_model = Lasso(alpha=0.5)

lasso_model.fit(x_train1, y_train)
y_pred_lasso = lasso_model.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_lasso))
print('MAE is: ', mean_absolute_error(y_test, y_pred_lasso))
print('RMSE is: ', mean_squared_error(y_test, y_pred_lasso) ** 0.5)
```

#Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Gradient Boosting Regression
gb_model = GradientBoostingRegressor(random_state=42)

gb_model.fit(x_train1, y_train)
y_pred_gb = gb_model.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_gb))
print('MAE is: ', mean_absolute_error(y_test, y_pred_gb))
print('MSE is: ', mean_squared_error(y_test, y_pred_gb) ** 0.5)
```

#SVM

```python
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Support Vector Regression with scaled data
svr_model = SVR()

svr_model.fit(x_train1, y_train)
y_pred_svr = svr_model.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_svr))
print('MAE is: ', mean_absolute_error(y_test, y_pred_svr))
print('MSE is: ', mean_squared_error(y_test, y_pred_svr) ** 0.5)
```
